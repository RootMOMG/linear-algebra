# Pregunta 1: Ejes naturales de una transformación 3D

**Enunciado:** Un desarrollador aplica una transformación a un objeto 3D y necesita conocer sus ejes naturales. ¿Qué debe calcular?

**Opciones:**
- **A)** Los eigenvalores y eigenvectores de la transformación.
- **B)** El determinante para medir el cambio de escala del objeto.
- **C)** Un sistema de ecuaciones lineales que describa el objeto.
- **D)** La matriz inversa para revertir la transformación.

## Respuesta correcta: **A) Los eigenvalores y eigenvectores de la transformación**

---
## Explicación paso a paso

### Paso 1: ¿Qué son los ejes naturales?

Los **ejes naturales** (o ejes principales) de una transformación son las direcciones en el espacio a lo largo de las cuales la transformación actúa de la forma más simple posible: **solo escala** (estira o comprime) sin cambiar la dirección.

En otras palabras, si un vector está alineado con un eje natural, la transformación solo lo multiplica por un número, no lo rota ni lo deforma de otra manera.

### Paso 2: Relación con eigenvectores

Por definición, un **eigenvector** $\vec{v}$ de una matriz $A$ es un vector que satisface:

$$A\vec{v} = \lambda \vec{v}$$

Es decir: cuando aplicamos la transformación $A$ al vector $\vec{v}$, el resultado es el **mismo vector** multiplicado por un escalar $\lambda$. La dirección no cambia, solo la magnitud.

**Conclusión:** Los eigenvectores son exactamente las direcciones de los ejes naturales.

### Paso 3: Papel de los eigenvalores

El **eigenvalor** $\lambda$ asociado a cada eigenvector indica:

- Cuánto se escala (estira o comprime) el vector en esa dirección.
- $|\lambda| > 1$: estiramiento
- $|\lambda| < 1$: compresión
- $\lambda < 0$: inversión de dirección

### Paso 4: ¿Por qué las otras opciones no son correctas?

| Opción | Razón por la que no responde la pregunta |
|--------|------------------------------------------|
| **B) Determinante** | Mide el cambio global de volumen/área, no las direcciones de los ejes naturales. |
| **C) Sistema de ecuaciones** | Describe restricciones del objeto, no los ejes principales de la transformación. |
| **D) Matriz inversa** | Sirve para deshacer la transformación, no para obtener sus direcciones principales. |

### Paso 5: Ejemplo numérico

Consideremos una matriz de transformación 3D. Calcularemos sus eigenvalores y eigenvectores para ver cómo definen los ejes naturales.

```python
import numpy as np

# Matriz de transformación 3D (ejemplo: escalado y algo de deformación)
A = np.array([
    [3, 0, 0],
    [0, 2, 0],
    [0, 0, 1]
])

print("Matriz de transformación A:")
print(A)
print()

# Calcular eigenvalores y eigenvectores
eigenvalores, eigenvectores = np.linalg.eig(A)

print("Eigenvalores (λ):", eigenvalores)
print()
print("Eigenvectores (columnas = ejes naturales):")
print(eigenvectores)
print()

# Verificar: A @ v = λ * v para cada par
print("Verificación A·v = λ·v:")
for i in range(3):
    v = eigenvectores[:, i].reshape(-1, 1)
    Av = A @ v
    lambda_v = eigenvalores[i] * v
    print(f"  Eigenvector {i+1}: A·v = {Av.flatten()}, λ·v = {lambda_v.flatten()} ✓")

print()
print("→ Los eigenvectores [1,0,0], [0,1,0], [0,0,1] SON los ejes naturales (X, Y, Z).")
print("→ Los eigenvalores 3, 2, 1 indican el factor de escala en cada eje.")
```

### Paso 6: Caso más general

Cuando la matriz incluye rotaciones o transformaciones más complejas, los eigenvectores ya no coinciden con los ejes estándar, pero siguen siendo las direcciones que solo se escalan.

```python
# Matriz con transformación más compleja (simétrica para que tenga eigenvectores reales)
B = np.array([
    [2, 1, 0],
    [1, 2, 0],
    [0, 0, 3]
])

eigenvalores_B, eigenvectores_B = np.linalg.eig(B)

print("Matriz B (transformación con mezcla en plano XY):")
print(B)
print()
print("Eigenvalores:", eigenvalores_B)
print("Eigenvectores (ejes naturales):")
for i in range(3):
    print(f"  Eje {i+1}: {eigenvectores_B[:, i]} (λ = {eigenvalores_B[i]:.2f})")
print()
print("→ Aunque B mezcla X e Y, los eigenvectores siguen siendo las direcciones
   que solo se escalan, no se rotan.")
```

---
## Resumen

| Concepto | Rol |
|----------|-----|
| **Eigenvectores** | Definen las **direcciones** de los ejes naturales (no cambian bajo la transformación, salvo escala). |
| **Eigenvalores** | Indican el **factor de escala** en cada eje natural. |

**Respuesta: A) Los eigenvalores y eigenvectores de la transformación.**
