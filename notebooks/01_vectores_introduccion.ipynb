## 1. ¿Qué es un Vector?

Un **vector** es una lista ordenada de números. En el contexto de álgebra lineal:

- Un vector en $\mathbb{R}^2$ tiene 2 componentes: $\vec{v} = (v_1, v_2)$
- Un vector en $\mathbb{R}^3$ tiene 3 componentes: $\vec{v} = (v_1, v_2, v_3)$
- En general, un vector en $\mathbb{R}^n$ tiene $n$ componentes

Los vectores pueden representar:
- Posiciones en el espacio
- Direcciones y magnitudes
- Datos (cada componente es una característica)

## 2. Representación de Vectores en Python

```python
# Creando vectores con NumPy

# Vector en R^2
v2 = np.array([3, 4])
print(f"Vector en R²: {v2}")
print(f"Dimensión: {v2.shape}")

# Vector en R^3
v3 = np.array([1, 2, 3])
print(f"\nVector en R³: {v3}")
print(f"Dimensión: {v3.shape}")
```

```python
# Vectores columna vs vectores fila

# Vector fila (1xn)
v_fila = np.array([[1, 2, 3]])
print(f"Vector fila: {v_fila}")
print(f"Shape: {v_fila.shape}")

# Vector columna (nx1)
v_columna = np.array([[1], [2], [3]])
print(f"\nVector columna:\n{v_columna}")
print(f"Shape: {v_columna.shape}")
```

## 3. Operaciones Básicas con Vectores

### 3.1 Suma de Vectores

La suma de vectores se realiza componente a componente:

$$\vec{u} + \vec{v} = (u_1 + v_1, u_2 + v_2, ..., u_n + v_n)$$

```python
# Suma de vectores
u = np.array([1, 2])
v = np.array([3, 1])

suma = u + v
print(f"u = {u}")
print(f"v = {v}")
print(f"u + v = {suma}")
```

### 3.2 Multiplicación por Escalar

Multiplicar un vector por un número (escalar) multiplica cada componente:

$$c \cdot \vec{v} = (c \cdot v_1, c \cdot v_2, ..., c \cdot v_n)$$

```python
# Multiplicación por escalar
v = np.array([2, 3])
c = 2

resultado = c * v
print(f"v = {v}")
print(f"c = {c}")
print(f"c * v = {resultado}")
```

### 3.3 Magnitud (Norma) de un Vector

La magnitud o norma de un vector $\vec{v}$ se calcula como:

$$||\vec{v}|| = \sqrt{v_1^2 + v_2^2 + ... + v_n^2}$$

```python
# Magnitud de un vector
v = np.array([3, 4])

# Método 1: Manual
magnitud_manual = np.sqrt(np.sum(v**2))

# Método 2: Usando np.linalg.norm
magnitud = np.linalg.norm(v)

print(f"v = {v}")
print(f"||v|| = {magnitud}")
```

### 3.4 Producto Punto (Producto Escalar)

El producto punto de dos vectores es un escalar:

$$\vec{u} \cdot \vec{v} = u_1 v_1 + u_2 v_2 + ... + u_n v_n$$

```python
# Producto punto
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

# Método 1: np.dot
producto = np.dot(u, v)

# Método 2: Operador @
producto_at = u @ v

print(f"u = {u}")
print(f"v = {v}")
print(f"u · v = {producto}")
```

## 4. Visualización de Vectores en 2D

```python
def plot_vectors(vectors, colors=None, labels=None):
    """
    Función para graficar vectores en 2D
    """
    fig, ax = plt.subplots(figsize=(8, 8))
    
    if colors is None:
        colors = plt.cm.tab10(np.linspace(0, 1, len(vectors)))
    
    for i, v in enumerate(vectors):
        color = colors[i] if isinstance(colors, (list, np.ndarray)) else colors
        label = labels[i] if labels else f'v{i+1}'
        ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, 
                  color=color, label=label)
    
    # Configurar límites del gráfico
    all_coords = np.array(vectors)
    max_val = np.max(np.abs(all_coords)) + 1
    ax.set_xlim(-max_val, max_val)
    ax.set_ylim(-max_val, max_val)
    
    ax.axhline(y=0, color='k', linewidth=0.5)
    ax.axvline(x=0, color='k', linewidth=0.5)
    ax.grid(True)
    ax.set_aspect('equal')
    ax.legend()
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    plt.title('Vectores en R²')
    plt.show()
```

```python
# Visualizar suma de vectores
u = np.array([2, 1])
v = np.array([1, 3])
w = u + v

plot_vectors([u, v, w], 
             colors=['blue', 'red', 'green'],
             labels=['u', 'v', 'u + v'])
```

```python
# Visualizar multiplicación por escalar
v = np.array([2, 1])

plot_vectors([v, 2*v, -1*v, 0.5*v], 
             colors=['blue', 'red', 'green', 'orange'],
             labels=['v', '2v', '-v', '0.5v'])
```
