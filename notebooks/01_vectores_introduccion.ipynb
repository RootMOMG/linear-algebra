## 1. ¿Qué es un Vector?

Un **vector** es una lista ordenada de números. En el contexto de álgebra lineal:

- Un vector en $\mathbb{R}^2$ tiene 2 componentes: $\vec{v} = (v_1, v_2)$
- Un vector en $\mathbb{R}^3$ tiene 3 componentes: $\vec{v} = (v_1, v_2, v_3)$
- En general, un vector en $\mathbb{R}^n$ tiene $n$ componentes

Los vectores pueden representar:
- Posiciones en el espacio
- Direcciones y magnitudes
- Datos (cada componente es una característica)

## 2. Representación de Vectores en Python

```python
# Creando vectores con NumPy

# Vector en R^2
v2 = np.array([3, 4])
print(f"Vector en R²: {v2}")
print(f"Dimensión: {v2.shape}")

# Vector en R^3
v3 = np.array([1, 2, 3])
print(f"\nVector en R³: {v3}")
print(f"Dimensión: {v3.shape}")
```

```python
# Vectores columna vs vectores fila

# Vector fila (1xn)
v_fila = np.array([[1, 2, 3]])
print(f"Vector fila: {v_fila}")
print(f"Shape: {v_fila.shape}")

# Vector columna (nx1)
v_columna = np.array([[1], [2], [3]])
print(f"\nVector columna:\n{v_columna}")
print(f"Shape: {v_columna.shape}")
```

## 3. Operaciones Básicas con Vectores

### 3.1 Suma de Vectores

La suma de vectores se realiza componente a componente:

$$\vec{u} + \vec{v} = (u_1 + v_1, u_2 + v_2, ..., u_n + v_n)$$

```python
# Suma de vectores
u = np.array([1, 2])
v = np.array([3, 1])

suma = u + v
print(f"u = {u}")
print(f"v = {v}")
print(f"u + v = {suma}")
```

### 3.2 Multiplicación por Escalar

Multiplicar un vector por un número (escalar) multiplica cada componente:

$$c \cdot \vec{v} = (c \cdot v_1, c \cdot v_2, ..., c \cdot v_n)$$

```python
# Multiplicación por escalar
v = np.array([2, 3])
c = 2

resultado = c * v
print(f"v = {v}")
print(f"c = {c}")
print(f"c * v = {resultado}")
```

### 3.3 Magnitud (Norma) de un Vector

La magnitud o norma de un vector $\vec{v}$ se calcula como:

$$||\vec{v}|| = \sqrt{v_1^2 + v_2^2 + ... + v_n^2}$$

```python
# Magnitud de un vector
v = np.array([3, 4])

# Método 1: Manual
magnitud_manual = np.sqrt(np.sum(v**2))

# Método 2: Usando np.linalg.norm
magnitud = np.linalg.norm(v)

print(f"v = {v}")
print(f"Magnitud manual: {magnitud_manual}")
print(f"||v|| = {magnitud}")
```

### 3.4 Producto Punto (Producto Escalar)

El producto punto de dos vectores es un escalar:

$$\vec{u} \cdot \vec{v} = u_1 v_1 + u_2 v_2 + ... + u_n v_n$$

```python
# Producto punto
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

# Método 1: np.dot
producto = np.dot(u, v)

# Método 2: Operador @
producto_at = u @ v

print(f"u = {u}")
print(f"v = {v}")
print(f"u · v = {producto}")
```

```python
# Bloque 1 — Vectores
# u · v = u₁·v₁ + u₂·v₂ + u₃·v₃

u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

print("Vectores:")
print(f"  u = {u}  →  u₁={u[0]}, u₂={u[1]}, u₃={u[2]}")
print(f"  v = {v}  →  v₁={v[0]}, v₂={v[1]}, v₃={v[2]}")
```

```python
# Bloque 2 — Paso 1: Productos componente a componente

p1 = u[0] * v[0]
p2 = u[1] * v[1]
p3 = u[2] * v[2]

print("Paso 1 — Productos componente a componente:")
print(f"  u₁·v₁ = {u[0]} × {v[0]} = {p1}")
print(f"  u₂·v₂ = {u[1]} × {v[1]} = {p2}")
print(f"  u₃·v₃ = {u[2]} × {v[2]} = {p3}")
```

```python
# Bloque 3 — Paso 2: Suma de los productos

producto_manual = p1 + p2 + p3

print("Paso 2 — Suma de los productos:")
print(f"  u·v = u₁·v₁ + u₂·v₂ + u₃·v₃")
print(f"  u·v = {p1} + {p2} + {p3} = {producto_manual}")
```

```python
# Bloque 4 — Resultado con validación

print(f"Resultado: u · v = {producto_manual}")
print(f"Verificación con np.dot(u, v): {np.dot(u, v)}")
```

### Producto punto igual a 0

- En 2D o 3D: forman un ángulo de 90°.
- En general: no hay “proyección” de uno sobre el otro; la suma de productos componente a componente se cancela.

**Ejemplo:** u = (1, 0) y v = (0, 1) → u·v = 1·0 + 0·1 = 0.

```python
# Explicación de producto punto igual a 0 (en código)

print("=" * 60)
print("EXPLICACIÓN: ¿Qué significa u · v = 0?")
print("=" * 60)
print()
print("Cuando u · v = 0, los vectores son ORTOGONALES (perpendiculares).")
print()
print("• En 2D o 3D: forman un ángulo de 90°.")
print("• En general: no hay 'proyección' de uno sobre el otro;")
print("  la suma de productos componente a componente se cancela.")
print()
print("Ejemplo numérico:")
print("-" * 40)

a = np.array([1, 0])
b = np.array([0, 1])

print(f"  a = {a}   b = {b}")
print(f"  a·b = {a[0]}·{b[0]} + {a[1]}·{b[1]} = {a[0]*b[0]} + {a[1]*b[1]} = {np.dot(a, b)}")
print()
print("→ a · b = 0  ⇒  a y b son perpendiculares (ortogonales).")
```

```python
# Explicación gráfica: vectores ortogonales (u · v = 0)

# a y b ya definidos en la celda anterior: a=(1,0), b=(0,1)
fig, ax = plt.subplots(figsize=(6, 6))

ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='C0', linewidth=2, label=f'a = {a}')
ax.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='C1', linewidth=2, label=f'b = {b}')

ax.set_xlim(-0.5, 1.5)
ax.set_ylim(-0.5, 1.5)
ax.axhline(0, color='k', linewidth=0.5)
ax.axvline(0, color='k', linewidth=0.5)
ax.grid(True, alpha=0.5)
ax.set_aspect('equal')
ax.legend(loc='upper right')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Producto punto = 0 → vectores perpendiculares (90°)')
plt.tight_layout()
plt.show()
```

## 4. Visualización de Vectores en 2D

```python
def plot_vectors(vectors, colors=None, labels=None):
    """
    Función para graficar vectores en 2D
    """
    fig, ax = plt.subplots(figsize=(8, 8))
    
    if colors is None:
        colors = plt.cm.tab10(np.linspace(0, 1, len(vectors)))
    
    for i, v in enumerate(vectors):
        color = colors[i] if isinstance(colors, (list, np.ndarray)) else colors
        label = labels[i] if labels else f'v{i+1}'
        ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, 
                  color=color, label=label)
    
    # Configurar límites del gráfico
    all_coords = np.array(vectors)
    max_val = np.max(np.abs(all_coords)) + 1
    ax.set_xlim(-max_val, max_val)
    ax.set_ylim(-max_val, max_val)
    
    ax.axhline(y=0, color='k', linewidth=0.5)
    ax.axvline(x=0, color='k', linewidth=0.5)
    ax.grid(True)
    ax.set_aspect('equal')
    ax.legend()
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    plt.title('Vectores en R²')
    plt.show()
```

```python
# Visualizar suma de vectores
u = np.array([2, 1])
v = np.array([1, 3])
w = u + v

plot_vectors([u, v, w], 
             colors=['blue', 'red', 'green'],
             labels=['u', 'v', 'u + v'])
```

```python
# Visualizar multiplicación por escalar
v = np.array([2, 1])

plot_vectors([v, 2*v, -1*v, 0.5*v], 
             colors=['blue', 'red', 'green', 'orange'],
             labels=['v', '2v', '-v', '0.5v'])
```

# Explicación de producto punto igual a 0

Cuando **u · v = 0**, los vectores son **ortogonales** (perpendiculares).

- **En 2D o 3D:** forman un ángulo de 90°.
- **En general:** no hay proyección de uno sobre el otro; la suma de productos componente a componente se cancela.

**Ejemplo:** u = (1, 0) y v = (0, 1) → u·v = 1·0 + 0·1 = 0.


```python
# Demostración gráfica de los vectores a y b en R³ + producto punto

from mpl_toolkits.mplot3d import Axes3D
from matplotlib import animation

# a y b definidos en celdas anteriores: a = [2, -1, 3], b = [-1, 4, 2]

# Producto punto y proyección de b sobre a: proj_a(b) = (a·b / |a|²) * a
a_dot_b = np.dot(a, b)
proj_b_sobre_a = (a_dot_b / np.dot(a, a)) * a

fig = plt.figure(figsize=(9, 8))
ax = fig.add_subplot(111, projection='3d')

# Vector a (azul), vector b (naranja), proyección de b sobre a (verde)
ax.quiver(0, 0, 0, a[0], a[1], a[2], color='C0', arrow_length_ratio=0.12, linewidth=2.5, label=f'a = {a}')
ax.quiver(0, 0, 0, b[0], b[1], b[2], color='C1', arrow_length_ratio=0.12, linewidth=2.5, label=f'b = {b}')
ax.quiver(0, 0, 0, proj_b_sobre_a[0], proj_b_sobre_a[1], proj_b_sobre_a[2],
          color='green', arrow_length_ratio=0.12, linewidth=2, linestyle='--', alpha=0.9,
          label=f'proyección de b sobre a (a·b/|a|²)·a')

# Segmento desde la punta de la proyección hasta la punta de b (componente perpendicular)
ax.plot([proj_b_sobre_a[0], b[0]], [proj_b_sobre_a[1], b[1]], [proj_b_sobre_a[2], b[2]],
       'k--', alpha=0.6, linewidth=1.5)

max_val = max(np.max(np.abs(a)), np.max(np.abs(b)), np.max(np.abs(proj_b_sobre_a))) + 1
ax.set_xlim(-max_val, max_val)
ax.set_ylim(-max_val, max_val)
ax.set_zlim(-max_val, max_val)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.legend(loc='upper left', fontsize=8)
ax.set_title(f'Vectores a, b y producto punto  →  a·b = {a_dot_b}')

def rotar_vista(frame):
    ax.view_init(elev=20, azim=frame)

anim = animation.FuncAnimation(fig, rotar_vista, frames=np.arange(0, 360, 2), interval=50, blit=False)
plt.tight_layout()

# Mostrar animación en el notebook (reproduce en el navegador)
from IPython.display import HTML
HTML(anim.to_jshtml())
```

## Aplicaciones reales del producto punto

El producto punto **u · v** mide cuánto se parecen dos vectores en dirección. Algunos usos concretos:

**1. Física — Trabajo:** El trabajo de una fuerza F al mover en dirección d es W = F · d. Solo cuenta la componente de F en la dirección del movimiento.

**2. Gráficos 3D y videojuegos — Iluminación:** La luz en una superficie depende del ángulo entre la normal y la dirección de la luz; se usa el coseno de ese ángulo (producto punto de vectores unitarios).

**3. Machine learning y búsqueda — Similitud:** La similitud coseno entre vectores (documentos, embeddings) es el producto punto de vectores normalizados. En búsqueda y recomendación, las puntuaciones suelen ser productos punto entre consulta e ítems.

**4. Proyecciones:** La proyección de v sobre u es (u·v/|u|²)·u. Se usa en mínimos cuadrados, filtros de señal y compresión.

**5. Perpendicularidad:** Si u · v = 0, los vectores son perpendiculares. Útil en colisiones, geometría y física.

**Resumen:** El producto punto sirve para alineación, proyecciones, trabajo, iluminación y similitud en física, gráficos, ML y geometría.

## Ejemplo: Análisis de ventas en un marketplace (producto punto por categoría y cliente)

Cada **cliente** se representa como un vector: las componentes son las ventas (o unidades) en cada **categoría**. Así podemos usar el **producto punto** (o la similitud coseno) para medir qué clientes tienen gustos parecidos o qué categorías se compran juntas.

```python
# Marketplace: ventas por categoría y por cliente
import numpy as np

# Categorías del marketplace
categorias = np.array(['Electrónica', 'Ropa', 'Hogar', 'Deportes', 'Libros'])

# Matriz de ventas: filas = clientes, columnas = categorías (unidades vendidas)
ventas = np.array([
    [120,  30,  45,  20,  80],   # Cliente A
    [ 90,  85,  25,  40,  60],   # Cliente B
    [110,  25,  50,  35,  75],   # Cliente C (similar a A)
    [ 15,  70,  90,  55,  20],   # Cliente D
    [ 20,  80,  85,  50,  25],   # Cliente E (similar a D)
])
clientes = np.array(['Cliente A', 'Cliente B', 'Cliente C', 'Cliente D', 'Cliente E'])

print('Ventas por cliente (filas) y categoría (columnas):')
print('Columnas:', categorias)
print(ventas)
print()

# Cada fila es el vector "compras" del cliente → producto punto entre filas = similitud
def similitud_coseno(u, v):
    n = np.linalg.norm(u) * np.linalg.norm(v)
    return (np.dot(u, v) / n) if n > 0 else 0.0

print('Similitud entre clientes (producto punto de vectores normalizados):')
for i in range(len(clientes)):
    for j in range(i + 1, len(clientes)):
        s = similitud_coseno(ventas[i], ventas[j])
        print(f'  {clientes[i]} vs {clientes[j]}: {s:.3f}')
print()

# Por categoría: cada columna es un vector (ventas por cliente) → correlación entre categorías
print('Similitud entre categorías (qué tan parecido es el perfil de compra):')
for i in range(len(categorias)):
    for j in range(i + 1, len(categorias)):
        u, v = ventas[:, i], ventas[:, j]
        s = similitud_coseno(u, v)
        print(f'  {categorias[i]} vs {categorias[j]}: {s:.3f}')
print()

# Ejemplo: "relevancia" de un cliente para un perfil ideal (producto punto)
perfil_objetivo = np.array([100, 50, 40, 30, 70])  # ej: más electrónica y libros
print('Puntuación de cada cliente respecto al perfil objetivo (producto punto con perfil):')
for i, nombre in enumerate(clientes):
    p = np.dot(ventas[i], perfil_objetivo)
    print(f'  {nombre}: {p}')
print('  → Cliente con mayor puntuación es el más alineado con ese perfil.')
```

## Ejemplo práctico: producto punto en apps de pedidos (comida a domicilio y envío de regalos)

En una app de **pedidos a domicilio** (comida y regalos), cada usuario puede representarse como un vector: componentes = pedidos por categoría. El **producto punto** sirve para:
- Encontrar usuarios con gustos similares (recomendaciones).
- Puntuar qué regalo o promoción encaja mejor con el perfil del destinatario.

```python
# App de pedidos: comida a domicilio + envío de regalos
import numpy as np

# Categorías de COMIDA en la app
categorias_comida = np.array(['Pizza', 'Sushi', 'Healthy', 'Postres', 'Comida rápida'])
# Categorías de REGALOS que se pueden enviar
categorias_regalos = np.array(['Flores', 'Gourmet', 'Dulces', 'Vinos', 'Experiencias'])

# Pedidos por usuario (filas) en cada categoría de comida (columnas)
pedidos_comida = np.array([
    [12,  8,  2,  5, 15],   # Ana - mucho rápido y pizza
    [ 5, 20, 10,  3,  2],   # Luis - sushi y healthy
    [14,  6,  1,  8, 14],   # María - similar a Ana
    [ 2, 18, 12,  2,  1],   # Pedro - sushi y healthy (como Luis)
    [ 8,  4,  4, 12,  9],   # Carla - variado, muchos postres
])
usuarios = np.array(['Ana', 'Luis', 'María', 'Pedro', 'Carla'])

# Historial de regalos enviados/recibidos por categoría (mismo orden de usuarios)
regalos_perfil = np.array([
    [1, 3, 4, 2, 0],   # Ana: dulces y gourmet
    [2, 4, 1, 5, 2],   # Luis: gourmet, vinos, experiencias
    [3, 2, 5, 1, 1],   # María: dulces y flores
    [1, 5, 0, 4, 3],   # Pedro: gourmet y vinos
    [4, 2, 4, 2, 0],   # Carla: flores y dulces
])

print('=== CONSUMO DE COMIDA (entregas en línea) ===')
print('Pedidos por usuario y categoría de comida:')
print('Columnas:', categorias_comida)
print(pedidos_comida)
print()

def similitud_coseno(u, v):
    n = np.linalg.norm(u) * np.linalg.norm(v)
    return (np.dot(u, v) / n) if n > 0 else 0.0

print('Similitud entre usuarios (producto punto normalizado) → recomendaciones:')
for i in range(len(usuarios)):
    for j in range(i + 1, len(usuarios)):
        s = similitud_coseno(pedidos_comida[i], pedidos_comida[j])
        print(f'  {usuarios[i]} vs {usuarios[j]}: {s:.3f}')
print()

print('=== ENVÍO DE REGALOS ===')
print('Perfil de regalos (envíos/gustos) por usuario. Columnas:', categorias_regalos)
print(regalos_perfil)
print()

# Catálogo de regalos: cada regalo = vector de "peso" por categoría
regalo_flores    = np.array([5, 0, 0, 0, 0])
regalo_gourmet  = np.array([0, 5, 0, 2, 0])
regalo_dulces   = np.array([0, 0, 5, 0, 0])
regalo_vinos    = np.array([0, 2, 0, 5, 1])
regalo_experiencia = np.array([0, 1, 0, 1, 5])
regalos_catalogo = [regalo_flores, regalo_gourmet, regalo_dulces, regalo_vinos, regalo_experiencia]
nombres_regalos = ['Ramo flores', 'Caja gourmet', 'Box dulces', 'Caja vinos', 'Experiencia']

print('Puntuación producto punto: regalo · perfil_destinatario (mejor match para cada usuario):')
for i, nombre in enumerate(usuarios):
    perfil = regalos_perfil[i]
    puntuaciones = [np.dot(perfil, r) for r in regalos_catalogo]
    mejor = np.argmax(puntuaciones)
    print(f'  {nombre}: mejor regalo = {nombres_regalos[mejor]} (puntos {puntuaciones[mejor]:.0f})')
print('  → Así la app sugiere qué regalo enviar a cada destinatario usando el producto punto.')
```
